```python
#Reading the csv file
import pandas as pd
from pandas import read_csv
#plotting the graph 
from matplotlib import pyplot as plt
#train test split
from sklearn.model_selection import train_test_split
#validation curve
from sklearn.model_selection import cross_val_score
#importing the KFold class
from sklearn.model_selection import StratifiedKFold
#importing the logistic regression model
from sklearn.linear_model import LogisticRegression
#importing the KNN classifier 
from sklearn.neighbors import KNeighborsClassifier
```

```python
path="C:\\Users\\user\\Downloads\\iris.csv"
names=["sepal-length","sepal-width","petal-length","petal-width","class"]
dataset=read_csv(path,names=names)
print(dataset)
#split the dataset into training,testing and validation
array=dataset.values
x=array[:,0:4]
y=array[:,4]
x_train,x_validation,y_train,y_validation=train_test_split(x,y,test_size=0.20,random_state=1,shuffle=True)
```

```python
#empty list for the models
models=[]
#models.append(('LR',Logistic Regression(solver,'liblinear')))
models.append(('LR',LogisticRegression()))
models.append(('KNN',KNeighborsClassifier()))
print(models)
result=[]
names=[]
for name,model in models:
    kfold=StratifiedKFold(n_splits=10,random_state=1,shuffle=True)
    cv_result=cross_val_score(model,x_train,y_train,cv=kfold,scoring='accuracy')
    result.append(cv_result)
    names.append(name)
    print
```
